```python
# ============================================================
# FEATURE AUDIT
# ============================================================

import pandas as pd
import numpy as np

tx_features = pd.read_csv("C:/Users/koush/OneDrive/Documents/Financial_Forensics/data/raw/eliptic++/txs_features.csv")

print("Transaction feature dataset loaded successfully.")
print("Shape:", tx_features.shape)

tx_features.head()
```

```python
# Feature columns

feature_cols = [
    col for col in tx_features.columns
    if col != "txId"
]

print("Total columns:", len(tx_features.columns))
print("Identifier column: txId")
print("Feature columns:", len(feature_cols))

print("\nFirst 10 feature columns:")
print(feature_cols[:10])

print("\nLast 10 feature columns:")
print(feature_cols[-10:])
```

```python
# ============================================================
# DATA TYPE AUDIT
# ============================================================

print("Data types of all columns:")
print(tx_features.dtypes.value_counts())

print("\nNon-numeric columns:")
non_numeric_cols = tx_features.select_dtypes(
    exclude=np.number
).columns.tolist()

print(non_numeric_cols)

print("\nNumeric columns:")
numeric_cols = tx_features.select_dtypes(
    include=np.number
).columns.tolist()

print("Number of numeric columns:", len(numeric_cols))
```

```python
# ============================================================
# MISSING VALUE AUDIT
# ============================================================

missing_count = tx_features[feature_cols].isnull().sum()

missing_summary = pd.DataFrame({
    "Feature": missing_count.index,
    "Missing_Count": missing_count.values,
    "Missing_Percentage": (
        missing_count.values / len(tx_features) * 100
    )
})

missing_summary = missing_summary.sort_values(
    "Missing_Count",
    ascending=False
)

print("Total missing values:",
      missing_summary["Missing_Count"].sum())

print("\nFeatures containing missing values:")

display(
    missing_summary[
        missing_summary["Missing_Count"] > 0
    ]
)
```

```python
# ============================================================
# CONSTANT / NEAR-CONSTANT FEATURE AUDIT
# ============================================================

constant_features = []

near_constant_features = []

for col in feature_cols:

    value_counts = tx_features[col].value_counts(
        normalize=True,
        dropna=False
    )

    unique_values = tx_features[col].nunique(
        dropna=True
    )

    top_percentage = (
        value_counts.iloc[0] * 100
        if len(value_counts) > 0
        else 0
    )

    # Completely constant
    if unique_values <= 1:

        constant_features.append({
            "Feature": col,
            "Unique_Values": unique_values,
            "Top_Value_Percentage": top_percentage
        })

    # >= 99.5% same value
    elif top_percentage >= 99.5:

        near_constant_features.append({
            "Feature": col,
            "Unique_Values": unique_values,
            "Top_Value_Percentage": top_percentage
        })


constant_df = pd.DataFrame(constant_features)

near_constant_df = pd.DataFrame(
    near_constant_features
)

print("Number of constant features:",
      len(constant_df))

print("Number of near-constant features:",
      len(near_constant_df))


print("\nConstant features:")
display(constant_df)


print("\nNear-constant features:")
display(near_constant_df)
```

```python
# ============================================================
# INVESTIGATE MISSING AUGMENTED FEATURES
# ============================================================

augmented_cols = [
    col for col in tx_features.columns
    if col in [
        "in_txs_degree",
        "out_txs_degree",
        "total_BTC",
        "fees",
        "size",
        "num_input_addresses",
        "num_output_addresses",
        "in_BTC_min",
        "in_BTC_max",
        "in_BTC_mean",
        "in_BTC_median",
        "in_BTC_total",
        "out_BTC_min",
        "out_BTC_max",
        "out_BTC_mean",
        "out_BTC_median",
        "out_BTC_total"
    ]
]

print("Elliptic++ augmented features found:")
print(augmented_cols)

print("\nNumber of augmented features:",
      len(augmented_cols))


# ------------------------------------------------------------
# Find transactions where augmented features are missing
# ------------------------------------------------------------

missing_augmented = tx_features[
    tx_features[augmented_cols].isna().all(axis=1)
].copy()

print("\nTransactions missing ALL augmented features:",
      len(missing_augmented))


# ------------------------------------------------------------
# Check their time-step distribution
# ------------------------------------------------------------

print("\nTime-step distribution of missing transactions:")

missing_time_steps = (
    missing_augmented["Time step"]
    .value_counts()
    .sort_index()
)

display(missing_time_steps.to_frame(
    name="Missing_Transactions"
))


# ------------------------------------------------------------
# Compare with total transactions per time step
# ------------------------------------------------------------

total_by_time = (
    tx_features["Time step"]
    .value_counts()
    .sort_index()
)

time_comparison = pd.DataFrame({

    "Total_Transactions":
        total_by_time,

    "Missing_Augmented":
        missing_time_steps

}).fillna(0)

time_comparison["Missing_Percentage"] = (
    time_comparison["Missing_Augmented"]
    / time_comparison["Total_Transactions"]
    * 100
)

print("\nMissing augmented features by time step:")

display(time_comparison)
```

```python
# ============================================================
# MISSINGNESS VS TRANSACTION CLASS
# ============================================================

tx_classes = pd.read_csv(r"C:\Users\koush\OneDrive\Documents\Financial_Forensics\data\raw\eliptic++\txs_classes.csv")

print("Class dataset shape:", tx_classes.shape)

print("\nClass columns:")
print(tx_classes.columns.tolist())

print("\nFirst 5 rows:")
display(tx_classes.head())
```

```python
# ============================================================
# MERGE CLASS INFORMATION
# ============================================================

# Merge class information with transaction features
tx_audit = tx_features[["txId", "Time step"] + augmented_cols].copy()

tx_audit = tx_audit.merge(
    tx_classes,
    on="txId",
    how="left"
)

print("Merged audit dataset shape:", tx_audit.shape)

print("\nColumns:")
print(tx_audit.columns.tolist())
```

```python
# ============================================================
# COMPARE MISSING AUGMENTED FEATURES BY CLASS
# ============================================================

# A transaction is considered missing augmented data
# when all 17 augmented features are missing.

tx_audit["Augmented_Missing"] = (
    tx_audit[augmented_cols]
    .isna()
    .all(axis=1)
)

print("Missing augmented transactions:")
print(
    tx_audit["Augmented_Missing"]
    .value_counts()
)


# ------------------------------------------------------------
# Class distribution of missing transactions
# ------------------------------------------------------------

missing_by_class = pd.crosstab(
    tx_audit["class"],
    tx_audit["Augmented_Missing"]
)

print("\nClass vs augmented-feature missingness:")
display(missing_by_class)


# ------------------------------------------------------------
# Percentage missing within each class
# ------------------------------------------------------------

class_missing_pct = (
    tx_audit
    .groupby("class")["Augmented_Missing"]
    .agg(
        Total_Transactions="count",
        Missing_Augmented="sum"
    )
)

class_missing_pct["Missing_Percentage"] = (
    class_missing_pct["Missing_Augmented"]
    / class_missing_pct["Total_Transactions"]
    * 100
)

print("\nMissing percentage within each class:")

display(class_missing_pct)
```

```python
# ============================================================
# CELL 10 — FEATURE GROUP MAPPING
# ============================================================

# Define the feature groups based on the Elliptic++ structure

local_features = [f"Local_feature_{i}" for i in range(1, 94)]

aggregate_features = [f"Aggregate_feature_{i}" for i in range(1, 73)]

augmented_features = [
    "in_txs_degree",
    "out_txs_degree",
    "total_BTC",
    "fees",
    "size",
    "num_input_addresses",
    "num_output_addresses",
    "in_BTC_min",
    "in_BTC_max",
    "in_BTC_mean",
    "in_BTC_median",
    "in_BTC_total",
    "out_BTC_min",
    "out_BTC_max",
    "out_BTC_mean",
    "out_BTC_median",
    "out_BTC_total"
]

# Create mapping table
feature_groups = pd.DataFrame({
    "feature": (
        ["Time step"] +
        local_features +
        aggregate_features +
        augmented_features
    )
})

# Assign group
feature_groups["group"] = (
    ["Time step"] +
    ["Local"] * len(local_features) +
    ["Aggregate"] * len(aggregate_features) +
    ["Elliptic++ augmented"] * len(augmented_features)
)

# Original feature number
feature_groups["feature_number"] = range(1, 184)

# Reorder columns
feature_groups = feature_groups[
    ["feature_number", "feature", "group"]
]

display(feature_groups.head(10))

print("\nLast 20 features:")
display(feature_groups.tail(20))

print("\nFeature group counts:")
print(feature_groups["group"].value_counts())
```
